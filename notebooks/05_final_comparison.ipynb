# Final Comparison

Simple side by side comparison of coverage and interval width across the
three methods: no correction, fixed conformal prediction, adaptive conformal
prediction.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Enter the results we already computed

```python
methods = ["No correction", "Fixed conformal", "Adaptive conformal"]

coverage = [75.2, 78.0, 78.7]
width = [5152.03, 5387.22, 5630.05]

target_coverage = 80
```

## Coverage comparison

```python
plt.figure(figsize=(6, 4))
bars = plt.bar(methods, coverage, color=["gray", "steelblue", "seagreen"])
plt.axhline(target_coverage, color="red", linestyle="--", label="target (80%)")
plt.ylabel("empirical coverage (%)")
plt.title("Coverage: before vs after correction")
plt.legend()

for bar, val in zip(bars, coverage):
    plt.text(bar.get_x() + bar.get_width()/2, val + 0.5, str(val), ha="center")

plt.show()
```

## Interval width comparison

This is the cost side: wider intervals mean less precise, but more honest, forecasts.

```python
plt.figure(figsize=(6, 4))
bars = plt.bar(methods, width, color=["gray", "steelblue", "seagreen"])
plt.ylabel("average interval width")
plt.title("Interval width: before vs after correction")

for bar, val in zip(bars, width):
    plt.text(bar.get_x() + bar.get_width()/2, val + 30, str(val), ha="center")

plt.show()
```

## Reading these two charts together

The coverage chart shows both correction methods move closer to the 80% target,
with adaptive conformal prediction slightly ahead of fixed conformal prediction.

The width chart shows why: adaptive conformal prediction buys that extra coverage
by using wider intervals on average. Neither method is free, better coverage
comes at the cost of less sharp (wider) predictions.
